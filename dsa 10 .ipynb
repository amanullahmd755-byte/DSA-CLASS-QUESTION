```python
arr = []

n = int(input("Enter number of elements: "))

print("Enter elements:")
for i in range(n):
    num = int(input())
    arr.append(num)

max_element = arr[0]

for i in range(1, n):
    if arr[i] > max_element:
        max_element = arr[i]

print("Maximum element =", max_element)
```
