```python
# 6. postfix Expression Evaluation using stack

def evaluate_postfix(expression):
    stack = []
    
    for token in expression.split():
        if token.isdigit():
            stack.append(int(token))
        else:
            b = stack.pop()
            a = stack.pop()
            
            if token == '+':
                stack.append(a + b)
            elif token == '-':
                stack.append(a - b)
            elif token == '*':
                stack.append(a * b)
            elif token == '/':
                stack.append(a // b)  # integer division
                
    return stack.pop()

# Example
expr = "5 6 2 + * 12 4 / -"
print(evaluate_postfix(expr))  
```
