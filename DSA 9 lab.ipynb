```python
class Queue:
    def __init__(self):
        self.queue = []

    # Enqueue
    def enqueue(self, item):
        self.queue.append(item)
        print(item, "added to queue")

    # Dequeue
    def dequeue(self):
        if len(self.queue) == 0:
            print("Queue is empty")
        else:
            print(self.queue.pop(0), "removed from queue")

    # Display
    def display(self):
        print("Queue:", self.queue)


# Example usage
q = Queue()

q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

q.display()

q.dequeue()
q.display()
```
