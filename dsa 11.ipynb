```python
arr = []

n = int(input("Enter number of elements: "))

print("Enter elements:")
for i in range(n):
    num = int(input())
    arr.append(num)

search = int(input("Enter element to search: "))

found = False

for i in range(n):
    if arr[i] == search:
        print("Element found at position", i + 1)
        found = True
        break

if not found:
    print("Element not found")
```
