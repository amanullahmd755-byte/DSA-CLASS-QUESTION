```python
def demonstrate_row_major(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    
    print(f"Original {rows}x{cols} Matrix:")
    for row in matrix:
        print(row)
    
    # Row-Major Flattening
    row_major_list = []
    print("\nTracing Memory Allocation (Row-Major):")
    
    for i in range(rows):
        for j in range(cols):
            element = matrix[i][j]
            index = i * cols + j  # The Row-Major Formula
            row_major_list.append(element)
            print(f"Element [{i}][{j}] (Value: {element}) -> Memory Index: {index}")
            
    print("\nFinal Row-Major representation in memory:")
    print(row_major_list)

# Define a 3x3 Matrix
example_matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

demonstrate_row_major(example_matrix)
```
